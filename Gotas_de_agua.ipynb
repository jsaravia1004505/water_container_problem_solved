```python
# Libraries
import random
import numpy as np
```

```python
#
# Dado un array de números enteros positivos, donde cada uno
# representa unidades de bloques apilados, debemos calcular cuantas unidades
# de agua quedarán atrapadas entre ellos.
#
# - Ejemplo: Dado el array [4, 0, 3, 6, 1, 3].
#
#         ⏹
#         ⏹
#   ⏹💧💧⏹
#   ⏹💧⏹⏹💧⏹
#   ⏹💧⏹⏹💧⏹
#   ⏹💧⏹⏹⏹⏹
#
#   Representando bloque con ⏹︎ y agua con 💧, quedarán atrapadas 7 unidades
#   de agua. Suponemos que existe un suelo impermeable en la parte inferior
#   que retiene el agua.
#
```

```python
# generate random lenght of the array
Largo = random.randint(4, 12)
print("Largo: " + str(Largo) + "\n")
arreglo = [0] * Largo
#arreglo = [5,3,1,3,5,1,1,2,9,5,1]
arreglo = np.array(arreglo)
print("Empty Array: ")
print(arreglo)
print()

# Fill each position of array with random int numbers
for i in range(Largo):
    # print(str(random.randint(0, 10)) + " ")
    arreglo[i] = random.randint(0, 10)

print("Filled array: ")
print(arreglo)
print()
```

```python
# Matriz creation

F = max(arreglo)
C = len(arreglo)
print("Filas = " + str(F) + " Columnas = " + str(C) + "\n")

# Matriz de dimensión (C, F) tipo string
matriz = np.full((F, C), ".", dtype=str)

print("Empty Matrix: ")
print(matriz[::-1, :])
```
